4.Value Counts

In Pandas, value_counts() is a fundamental data exploration tool used to calculate the frequency distribution of unique items within a dataset. It essentially functions as a built-in histogram or tally sheet, summarizing how many times each distinct value appears.

syntax--- df["armor_type"].value_counts()


```python
df["Defence"].max()
```

```python
df["Strength"].max()
```

```python
df["Defence"].mean() # average defence
```

```python
df.sort_values("Defence", ascending = False).head(9)
```

```python
df["Category"].value_counts() # check all types of armors count
```

```python
df["Category"].value_counts(normalize= True) * 100  # it tell the persentage of all armor types
```

```python
df[df["Strength"] == df["Strength"].max()]    
```

1. df["strength"].max() → highest strength search
2. df["strength"] == ... → find where highest strength is 
3. df[...] → show the full row

```python
df[df["Defence"] == df["Defence"].max()]  
```
